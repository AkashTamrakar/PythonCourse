# Introduction of Tuple 

Explanation 

A tuple in Python is a built-in data collection type used to store multiple items in a single variable. It is ordered, meaning items have a defined sequence that won't change, and immutable, meaning you cannot alter, add, or remove elements once it is created.

short Tuple are the ordered colloection of item that are immutable, they are similar ti list but there immutablility makes them different

```python
# Creating a tuple

empty_tuple = ()  # An empty tuple
print(empty_tuple)
print(type(empty_tuple))  # Output: <class 'tuple'>

 
```

```python
lst = list()
print(type(lst))
tpl = tuple()
print(type(tpl))
```

```python
numbers = tuple([1, 2, 3, 4, 5]) #  conversion from list to tuple
numbers # A tuple with integers 
print(type(numbers))  # Output: <class 'tuple'>
print(numbers)  # Output: (1, 2, 3, 4, 5)
```

```python
# packing and unpacking Tuple

# packing 
packed_tuple = 1,"akash", 2.32
print(packed_tuple)
```

```python
# unpacking a tuple

a,b,c = packed_tuple

print(a)
print(b)
print(c)
```

```python
# unpacking with *

numbers =(1,2,3,4,5,6)

first,*middle,last = numbers

print(first)
print(middle)
print(last)
```

```python
nested_tuple = ((1,2,3),("a","b","c","d"),(True, False))

## Access the element inside the tuple

print(nested_tuple[0])
print(nested_tuple[0][2])
print(nested_tuple[2][0])
print(nested_tuple[1][3])
```

```python
## iterating nested tuple

for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item, end=" ")
    print()
```
