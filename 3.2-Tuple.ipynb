# Introduction of Tuple 

Explanation 

A tuple in Python is a built-in data collection type used to store multiple items in a single variable. It is ordered, meaning items have a defined sequence that won't change, and immutable, meaning you cannot alter, add, or remove elements once it is created.

short Tuple are the ordered colloection of item that are immutable, they are similar ti list but there immutablility makes them different

```python
# Creating a tuple

empty_tuple = ()  # An empty tuple
print(empty_tuple)
print(type(empty_tuple))  # Output: <class 'tuple'>

 
```

```python
lst = list()
print(type(lst))
tpl = tuple()
print(type(tpl))
```

```python
numbers = tuple([1, 2, 3, 4, 5]) #  conversion from list to tuple
numbers # A tuple with integers 
print(type(numbers))  # Output: <class 'tuple'>
print(numbers)  # Output: (1, 2, 3, 4, 5)
```

```python
lst=list((1, 2, 3, 4, 5)) #  coversion from tuple to list
print(type(lst))
print(lst)
```

```python
# mixed tuple

mixed_tuple = (1, "Hello", 3.14, True)
print(mixed_tuple)
```

```python
##  Accessing Tuple Elements

numbers
```

```python
numbers[0]  # Accessing the first element of the tuple
```

```python
numbers[3]  # Accessing the fourth element of the tuple
```

```python
numbers[-1]  # Accessing the last element of the tuple
```

```python
numbers[0:3]  # Accessing elements from index 0 to 2 (3 is exclusive)
```

```python
numbers[::-1]  # Accessing elements in reverse order
```

```python
# Tuple operations


numbers + mixed_tuple  # Concatenation of two tuples
```

```python
concatinated_tuple = numbers + mixed_tuple

print(concatinated_tuple)  # Output: (1, 2, 3, 4, 5, 1, 'Hello', 3.14, True)
```

```python
mixed_tuple * 2  # Repeating the tuple elements

```

```python
mixed_tuple * 3  # Repeating the tuple elements
```

```python
numbers * 3
```

```python
# immutable Nature of Tuple

# Tuple Immutable, meaning their element connot be change once assigned

lst = [1,2,3,4,5]
print(lst)

lst[1] = "Akash"
print(lst)
```
