```python
# To Do list 


to_do_list = ["Buy groceries", "Clean the house", "Finish the project", "Call mom", "Exercise"]

# add a new task to the list
to_do_list.append("Read a book")
to_do_list.append("Pay bills")

# remove a task from the list
to_do_list.remove("Clean the house")

# check if a task is in the list
if "Finish the project" in to_do_list:
    print("You have a task to finish the project.")
print("to do list remaining:")
for task in to_do_list:
    print(f"- {task}")
```

```python
#  Organize student grades

grades = [85, 92, 78, 90, 88, 95, 80]

 # Adding a new grade
grades.append(89)

# calculating the average grade
average_grade = sum(grades) / len(grades)
print(f"- Average grade is: {average_grade :.2f} ")

# find the higest and lowest grades
higest_grade = max(grades)
lowest_grade = min(grades)

print(f"- Higest grade is: {higest_grade}")
print(f"- lowest grade is: {lowest_grade}")
```
