```python
# Que1: Temprature Conversion

def convert_temp(temp, unit):
    if unit == 'c':
        return temp * 9/5 + 32 # Convert Celsius to Fahrenheit
    elif unit == 'f':
        return (temp -32) * 5/9  # Convert Fahrenheit to Celsius
    else : 
        return "Invalid unit. Please use 'c' for Celsius or 'f' for Fahrenheit."


```

```python
print(convert_temp(25, 'c'))
```

```python
print(convert_temp(125, 'f'))
```

```python
#  Password Stringth Checker 

def is_strong_password(password):
    if len(password) < 8:
        return False
    if not any(char.isdigit() for char in password): # Check for digits
        return False
    if not any(char.isupper() for char in password): # Check for uppercase letters
        return False
    if not any(char.islower() for char in password): # Check for lowercase letters
        return False
    if not any(char in "!@#$%^&*()-_=+[]{}|;:'\",.<>?/`~" for char in password): # Check for special characters
        return False
    return True


```

```python
print(is_strong_password("Akash@12345"))
```

```python
print(is_strong_password("Akash")) 
```

```python
# Calculate the total cost of items in a shopping cart,

def calculate_total_cost(cart):
    total_cost = 0
    for item in cart:
        total_cost += item["price"] * item["quantity"]
    return total_cost

cart = [
    {"name": "Apple", "price": 1.5, "quantity": 4},
        {"name": "Banana", "price": 0.5, "quantity": 6},
        {"name": "Orange", "price": 2.0, "quantity": 3},
        {"name": "Grapes", "price": 3.0, "quantity": 2} 

    ] # this is a placeholder for the cart items, you can replace it with actual items and their prices
```

```python
#  call the function and print the total cost

total_cost = calculate_total_cost(cart)
print(f"The total cost of items in the cart is: ${total_cost:.2f}")
```

```python
#    #  Check if string is palandrome or not

def is_palindrome(string):
    # -- Remove spaces and convert to lowercase for uniformity
    
    string = string.lower().replace(" ", "")
    return string == string[::-1]  # Check if the string is equal to its reverse

print(is_palindrome("A man a plan a canal Panama"))  # True
print(is_palindrome("Hello good morning"))  # False
```

```python
 ##  calculate the factorial of a number using recursion

def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n-1)

print(factorial(5))  # 120

print(factorial(0))  # 1

print(factorial(5))  # 1
```
