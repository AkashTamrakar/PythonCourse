```python
# Que1: Temprature Conversion

def convert_temp(temp, unit):
    if unit == 'c':
        return temp * 9/5 + 32 # Convert Celsius to Fahrenheit
    elif unit == 'f':
        return (temp -32) * 5/9  # Convert Fahrenheit to Celsius
    else : 
        return "Invalid unit. Please use 'c' for Celsius or 'f' for Fahrenheit."


```

```python
print(convert_temp(25, 'c'))
```

```python
print(convert_temp(125, 'f'))
```

```python
#  Password Stringth Checker 

def is_strong_password(password):
    if len(password) < 8:
        return False
    if not any(char.isdigit() for char in password): # Check for digits
        return False
    if not any(char.isupper() for char in password): # Check for uppercase letters
        return False
    if not any(char.islower() for char in password): # Check for lowercase letters
        return False
    if not any(char in "!@#$%^&*()-_=+[]{}|;:'\",.<>?/`~" for char in password): # Check for special characters
        return False
    return True


```

```python
print(is_strong_password("Akash@12345"))
```

```python
print(is_strong_password("Akash"))
```
