```python
#  even odd checker

def even_or_odd(num):
    # this function check even or odd number
    if num%2 == 0:
        print(f"{num} is even number")
    else:
        print(f"{num} is odd number")
```

```python
# calling the function

even_or_odd(24)
```

```python
# function with multiple parameters

def print_info(name, age, city):
    print(f"Name: {name}")
    print(f"Age: {age}")
    print(f"City: {city}")
    
```

```python
print_info("Akash", 25, "Bhopal")
```

```python
def add_numbers(a, b):
    return a + b
result = add_numbers(5, 10)
print(f"The sum is: {result}")
```

```python
#  Variable length arguments

# Positional and keyword arguments


# positional arguments are the most common type of arguments in Python functions. 
# They are passed to the function in the order in which they are defined. For example:

def print_function(*args): # this function can take any number of positional arguments
    for num in args:
        print(num)

print_function(1, 2, 3, 4, 5, 6, 7, "Akash")  # This will print all the numbers passed as arguments 
```

```python
#   keyworad arguments are passed to the function using the name of the parameter.

def print_details(**kwargs):
    for key, value in kwargs.items(): # item() method returns a view object that displays a list of a dictionary's key-value tuple pairs.   
        print(f"{key}: {value}")

    
```

```python
print_details(name="Akash", age=25, city="Bhopal")  # This will print the details passed as key and value arguments
```

```python
def print_details(*args, **kwargs):
    for val in args:
        print(f"Positional argument: {val}")

    for key, value in kwargs.items(): # item() method returns a view object that displays a list of a dictionary's key-value tuple pairs.   
        print(f"Keyword argument - {key}: {value}")
        
    
```

```python
print_details(1, 2, 3, name="Akash", age=25, city="Bhopal")  # This will print the details passed as key and value arguments along with positional arguments
```
