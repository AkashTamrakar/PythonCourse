# Dictionary 

Introduction 

Dictionary is the unordered collection of items, the item store in key value pare, key must be unique and immutable data type (string, numbers and tuple) while values can be anything (including list and dictionary)



```python
# creating dictionary

empty_dict = {}
print(empty_dict)
print(type(empty_dict))
```

```python
empty_dict = dict()
empty_dict
```

```python
# add Data in dictionary

student = {"name":"Akash", "Age":17, "Roll no": "0123"}
print(type(student))
print(student)
```

```python
# Accessing dictionary element 

print(student['Age'])

print(student['Roll no'])
```

```python
# Accessing using get() method

print(student.get('name'))


print(student.get('Age'))

print(student.get('grade'))

print(student.get('grade', "Not available"))

# update dictionary

```

```python
#  Modifing Dictionary method

#  Dictionary are mutable, so we can  add, update and delete element

print(student)
```

```python
# Update the age of student

student["Age"]=21
print(student)
```

```python
# add element in student data

student["Address"]="Bhopal"
print(student)
```

```python
# delete the key and value pair

del student['Roll no']

print(student)
```

```python
#  Dictionary Method

# get all the keys

keys=student.keys()
print(keys)
```

```python

# get all the values

values=student.values()
print(values)
```

```python
# get all the key value pairs

items=student.items()
print(items)
```

```python
# shallow Copy 

student_copy=student
print(student)
print(student_copy)
```

```python
student["name"]="Ritik"
print(student)
print(student_copy)
```

```python
student_copy1=student.copy() # shallow copy
print(student_copy1)
print(student)
```

```python
student["name"]="Aman"
print(student)
print(student_copy1)
```

```python
#  Iterating Over Dictionary

##  we can use loop to iterate over dictionararise, key, value or items

## iterating over  keys

for keys in student.keys():
    print(keys)
```

```python
##  iterate over values

for values in student.values():
    print(values)
```

```python
# #  iterate over key value pair

for key, value in student.items():
    print(f"{key}:{value}")
```

```python
## nested Dictionary 

student={
    "student1":{"name":"Akash", "Age":21},
    "student2":{"name":"Ritik", "Age":19}
}
print(student)
```

```python
## Access nested dictionary element

print(student["student2"]["Age"])
print(student["student1"]["name"])
```

```python
student.items()
```

```python
###  Iterating over nested dictionaries

for student_id, student_info in student.items():
    print(f"{student_id}:{student_info}")
    for key, value in student_info.items():
        print(f"{key}:{value}")
```

```python
## Dictionaries Comprehension

squares={x:x**2 for x in range(5)}
print(squares)
```

```python
## conditional Dictionaries comprehension 

even={x:x**2 for x in range(10) if x%2==0}
print(even)
```

```python
#####    Pratical Example


### use the dectionary to count he frequency of element in list

numbers=[1,2,2,3,3,3,4,4,4,4]
frequency={}

for number in numbers:
    if number in frequency:
        frequency[number]+=1
    else:
        frequency[number]=1
print(frequency)
```

```python
####    Merge 2 Dictionaries into one


dict1 = {"a":1,"b":2}
dict2 = {"b":3,"d":4}

merged_dict = {**dict1,**dict2}

print(merged_dict)
```
