# Dictionary 

Introduction 

Dictionary is the unordered collection of items, the item store in key value pare, key must be unique and immutable data type (string, numbers and tuple) while values can be anything (including list and dictionary)



```python
# creating dictionary

empty_dict = {}
print(empty_dict)
print(type(empty_dict))
```

```python
empty_dict = dict()
empty_dict
```

```python
# add Data in dictionary

student = {"name":"Akash", "Age":17, "Roll no": "0123"}
print(type(student))
print(student)
```

```python
# Accessing dictionary element 

print(student['Age'])

print(student['Roll no'])
```

```python
# Accessing using get() method

print(student.get('name'))


print(student.get('Age'))

print(student.get('grade'))

print(student.get('grade', "Not available"))

# update dictionary

```
