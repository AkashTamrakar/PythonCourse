```python
## Accessing List Elements

fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
print(fruits[0])  # Access the first element
print(fruits[2])  # Access the third element
print(fruits[-1])  # Access the last element
```

```python
print(fruits[1:4])
print(fruits[1:])
```

```python
## modifying list elements

fruits[1] = "blueberry"
print(fruits)
```

```python
fruits[1:] = "papaya"
```

```python
fruits
```

```python
## adding elements to a list

fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
```

```python
fruits.append("orange") # add an element to the end of the list
print(fruits)
```

```python
fruits.insert(1, "grapes") # add an element at a specific index
print(fruits)
```

```python
fruits.remove("kiwi") # remove an element by value
print(fruits)

```

```python
pop_fruit=fruits.pop() # remove the last element
print(fruits)
print("Popped fruit:", pop_fruit)
```
